```python
import json
from torchvision.datasets import STL10
from assignment_01.task1.config import task_config
from assignment_01.task1.data.download import prepare_conflict_dataset
from pathlib import Path

d = prepare_conflict_dataset(task_config.TASK_CONFLICT_DATASET_DIR)
labels = STL10(task_config.TASK_DATASET_DIR, split="test", download=False).labels
recs = [json.load(open(d / e["metadata_path"])) for e in json.load(open(d / "manifest.json"))]
bad = [r["conflict_id"] for r in recs
       if labels[r["content_id"]] != r["content_label"] or labels[r["style_id"]] != r["style_label"]]
print("mismatches:", len(bad), bad[:5])   # expect 0
```
